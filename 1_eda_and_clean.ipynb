```python
df = pd.read_excel('UChicago Carillon Library.xlsx')

df = df.dropna(subset=['Title'])
composers = df['Composer'].unique()

df_size = df.groupby(['Composer']).size()
df_size = df_size.reset_index(name='count').sort_values(by='count',ascending=False)

# Save composers to create intermediate tags
df_composer_tag = df_size[['Composer']]
df_composer_tag.to_csv("composers_list.csv", index=False)
```

```python
import altair as alt
alt.data_transformers.enable("vegafusion")
import pandas as pd

base = (
    alt.Chart(df_size)
    .transform_window(
        sort=[{"field": "count", "order": "descending"}],
        cum_count="sum(count)",
        row_num="row_number()",
        frame=[None, 0],
    )
    .transform_joinaggregate(total_count="sum(count)", total_rows="count()")
    .transform_calculate(
        cum_items_pct="datum.row_num / datum.total_rows",
        cum_count_pct="datum.cum_count / datum.total_count",
    )
)

# 2. Cumulative curve + points
curve = base.mark_line(point=True, color="#1f77b4").encode(
    x=alt.X(
        "cum_items_pct:Q",
        title="Cumulative % of Items",
        axis=alt.Axis(format="%"),
    ),
    y=alt.Y(
        "cum_count_pct:Q",
        title="Cumulative % of Total Count",
        axis=alt.Axis(format="%"),
    ),
    tooltip=[
        alt.Tooltip("Composer:N", title="Category"),
        alt.Tooltip("count:Q", title="Count"),
        alt.Tooltip("cum_items_pct:Q", title="Cum. Items", format=".1%"),
        alt.Tooltip("cum_count_pct:Q", title="Cum. Count", format=".1%"),
    ],
)

# 3. Diagonal equality reference line (n% = n%)
# ref_df = pd.DataFrame({"x":[], "y": [0, 1]})
# ref_line = (
#     alt.Chart(ref_df)
#     .mark_line(strokeDash=[5, 5], color="gray")
#     .encode(x="x:Q", y="y:Q")
# )

chart = (curve).properties(
    width=400, height=300, title="Cumulative Share (Pareto) Curve"
)
chart
```
